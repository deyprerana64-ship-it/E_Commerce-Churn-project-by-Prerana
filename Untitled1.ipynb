Let's visualize the pairwise relationships between numerical columns, colored by churn status.

```python
import itertools
from plotly.subplots import make_subplots
import plotly.graph_objects as go # Changed to directly use graph_objects
import math

# Generate all unique pairs of numerical columns for scatter plots
numeric_pairs = list(itertools.combinations(numeric_cols, 2))

num_cols_per_row = 3 # Adjust as needed for desired layout
num_rows = math.ceil(len(numeric_pairs) / num_cols_per_row)

# Create subplot titles based on the pairs
subplot_titles = [f'{pair[0]} vs {pair[1]}' for pair in numeric_pairs]

fig = make_subplots(rows=num_rows, cols=num_cols_per_row, subplot_titles=subplot_titles)

churn_labels_map = {0: 'Not Churned', 1: 'Churned'}
churn_colors_map = {0: 'blue', 1: 'orange'} # Define colors for consistency

for i, (col1, col2) in enumerate(numeric_pairs):
    row_subplot = (i // num_cols_per_row) + 1
    col_subplot = (i % num_cols_per_row) + 1

    # Iterate over unique churn values to create separate traces
    for churn_val, churn_label in churn_labels_map.items():
        # Filter data for the current churn status
        data_for_churn = raw_data[raw_data['Churn'] == churn_val]

        fig.add_trace(go.Scatter(
            x=data_for_churn[col1],
            y=data_for_churn[col2],
            mode='markers',
            name=churn_label, # Set trace name for legend
            marker=dict(color=churn_colors_map[churn_val]), # Set marker color
            showlegend=True # Always show legend for all traces, Plotly will deduplicate
        ), row=row_subplot, col=col_subplot)

    # Update axis labels for each subplot
    fig.update_xaxes(title_text=col1, row=row_subplot, col=col_subplot)
    fig.update_yaxes(title_text=col2, row=row_subplot, col=col_subplot)

# Update layout for the entire figure
fig.update_layout(height=400 * num_rows, width=1200,
                  title_text='Pairwise Scatter Plots of Numeric Columns by Churn Status',
                  showlegend=True) # Ensure a single legend for the entire figure
fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import math

num_cols_per_row = 2 # Number of violin plots per row
num_rows = math.ceil(len(numeric_cols) / num_cols_per_row)

subplot_titles = [f'Distribution of {col}' for col in numeric_cols]

fig = make_subplots(rows=num_rows, cols=num_cols_per_row, subplot_titles=subplot_titles)

churn_labels_map = {0: 'Not Churned', 1: 'Churned'}
churn_colors_map = {0: 'blue', 1: 'orange'}

for i, col_name in enumerate(numeric_cols):
    row_subplot = (i // num_cols_per_row) + 1
    col_subplot = (i % num_cols_per_row) + 1

    for churn_val, churn_label in churn_labels_map.items():
        # Filter data for the current churn status
        data_for_churn = raw_data[raw_data['Churn'] == churn_val]

        fig.add_trace(go.Violin(
            y=data_for_churn[col_name],
            x=[churn_label] * len(data_for_churn),
            name=churn_label,
            legendgroup=churn_label,
            showlegend=(i == 0),
            scalemode='count',
            line_color=churn_colors_map[churn_val],
            fillcolor=churn_colors_map[churn_val],
            opacity=0.9
        ), row=row_subplot, col=col_subplot)

    fig.update_xaxes(title_text='Churn Status', row=row_subplot, col=col_subplot)
    fig.update_yaxes(title_text=col_name, row=row_subplot, col=col_subplot)

fig.update_layout(height=400 * num_rows, width=1200,
                  title_text='Distribution of Numeric Columns by Churn Status (Violin Plots)',
                  violinmode='group',
                  showlegend=True)
fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.express as px
import math

# Assuming `raw_data` and `categorical_cols` are defined in previous cells.
# `categorical_cols` is expected to be ['Churn', 'Complain', 'MaritalStatus', 'PreferedOrderCat']

num_cols_per_row = 3
num_rows = math.ceil(len(categorical_cols) / num_cols_per_row)

fig = make_subplots(rows=num_rows, cols=num_cols_per_row, subplot_titles=categorical_cols)

churn_labels_map = {0: 'Not Churned', 1: 'Churned'}

for i, col_name in enumerate(categorical_cols):
  row_subplot = (i // num_cols_per_row) + 1
  col_subplot = (i % num_cols_per_row) + 1

  if col_name == 'Churn':
    # For the 'Churn' column itself, display its distribution with custom x-axis labels
    churn_counts = raw_data[col_name].value_counts().sort_index()
    fig.add_trace(go.Bar(x=[churn_labels_map[val] for val in churn_counts.index],
                         y=churn_counts.values,
                         name='Churn Distribution',
                         showlegend=False), # No need for a legend entry for this specific subplot
                    row=row_subplot, col=col_subplot)
    fig.update_xaxes(title_text='Churn Status', row=row_subplot, col=col_subplot)
  else:
    # For other categorical columns, create grouped bar charts by 'Churn' status
    temp_fig = px.histogram(raw_data, x=col_name, color='Churn',
                            barmode='group',
                            category_orders={'Churn': [0, 1]}, # Ensure consistent order for churn categories
                            labels={'Churn': 'Churn Status'}) # Label for the legend title

    for trace in temp_fig.data:
        # Update trace name for the legend (e.g., '0' -> 'Not Churned', '1' -> 'Churned')
        current_churn_value = int(trace.name)
        trace.name = churn_labels_map.get(current_churn_value, f'Unknown Churn ({current_churn_value})')
        fig.add_trace(trace, row=row_subplot, col=col_subplot)

    fig.update_xaxes(title_text=col_name, row=row_subplot, col=col_subplot)

# Update layout for the entire figure
fig.update_layout(height=400 * num_rows, width=1000,
                  title_text='Distribution of Categorical Columns by Churn Status',
                  showlegend=True, # Show legend for churn status (Not Churned, Churned)
                  barmode='group') # Set barmode for the entire figure to group bars
fig.show()
```

# MISSING VALUE **IMPUTATION**

```python
data_cleaned=raw_data.copy()
data_cleaned.isnull().sum()
```

```python
for col in numeric_cols:
  median_val=data_cleaned[col].median()
  data_cleaned[col]=data_cleaned[col].fillna(median_val)
print(data_cleaned.isnull().sum())
```

# OUTLIER CHECK

```python
for col in numeric_cols:
  q1=data_cleaned[col].quantile(0.25)
  q3=data_cleaned[col].quantile(0.75)
  iqr=q3-q1
  lower_bound=q1-1.5*iqr
  upper_bound=q3+1.5*iqr
  outliers=data_cleaned[(data_cleaned[col]<lower_bound)|(data_cleaned[col]>upper_bound)]
  print(f'No of outliers in column{col}:{len(outliers)}')
```

```python
for col in numeric_cols:
  q1=data_cleaned[col].quantile(0.25)
  q3=data_cleaned[col].quantile(0.75)
  iqr=q3-q1
  lower_bound=q1-1.5*iqr
  upper_bound=q3+1.5*iqr
  data_cleaned[col]=np.where(data_cleaned[col]<lower_bound,lower_bound,np.where(data_cleaned[col]>upper_bound,upper_bound,data_cleaned[col]))
print(data_cleaned[numeric_cols].describe())
```

```python
for col in numeric_cols:
  q1=data_cleaned[col].quantile(0.25)
  q3=data_cleaned[col].quantile(0.75)
  iqr=q3-q1
  lower_bound=q1-1.5*iqr
  upper_bound=q3+1.5*iqr
  outliers=data_cleaned[(data_cleaned[col]<lower_bound)|(data_cleaned[col]>upper_bound)]
  print(f'No of outliers in column{col}:{len(outliers)}')
```

# TRAIN TEST SPLIT

```python
x=data_cleaned.drop('Churn',axis=1)
y=data_cleaned['Churn']
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42,stratify=y)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# **Model** **fitting**

```python
categorical_cols=['PreferedOrderCat','MaritalStatus','Complain']
numeric_cols=['Tenure', 'WarehouseToHome', 'NumberOfDeviceRegistered','SatisfactionScore', 'NumberOfAddress','DaySinceLastOrder', 'CashbackAmount']
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
preprocessor=ColumnTransformer(transformers=[('num',StandardScaler(),numeric_cols),('cat',OneHotEncoder(handle_unknown='ignore'),categorical_cols)])

from sklearn.linear_model import LogisticRegression
from sklearn .pipeline import Pipeline
model=Pipeline([('preprocessor',preprocessor),('classifier',LogisticRegression(max_iter=1000,class_weight="balanced"))])
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(y_pred)
y_prob=model.predict_proba(x_test)
print(y_prob)
```

# Evaluation metrics

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report,roc_auc_score,RocCurveDisplay,roc_curve
accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
roc_auc=roc_auc_score(y_test,y_prob[:, 1])
confusion_matrix=confusion_matrix(y_test,y_pred)
classification_report=classification_report(y_test,y_pred)
print('Accuracy:',accuracy)
print('Precision:',precision)
print('Recall:',recall)
print('F1 Score:',f1)
print('ROC_AUC:',roc_auc)
print('Confusion Matrix:')
print(confusion_matrix)
print('Classification_report:')
print(classification_report)
RocCurveDisplay.from_predictions(y_test,y_prob[:,1])
plt.show()
```

```python
train_acc=model.score(x_train,y_train)
test_acc=model.score(x_test,y_test)
print('Train Accuracy:',train_acc)
print('Test Accuracy:',test_acc)
```
